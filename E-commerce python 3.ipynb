```python
import matplotlib.pyplot as plt

df['Sales'] = df['Quantity'] * df['Price']

product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

product_sales.plot(kind='bar', figsize=(10,5))

plt.title('Product-wise Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

category_sales.plot(kind='bar', figsize=(8,5))

plt.title('Category-wise Sales')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
city_sales = df.groupby('City')['Sales'].sum().sort_values(ascending=False)

city_sales.plot(kind='bar', figsize=(10,5))

plt.title('City-wise Sales')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Sales'].sum()

monthly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
category_quantity = df.groupby('Category')['Quantity'].sum().sort_values(ascending=False)

category_quantity.plot(kind='bar', figsize=(8,5))

plt.title('Category-wise Quantity Sold')
plt.xlabel('Category')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)
plt.show()
```
